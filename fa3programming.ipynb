```python
# fa3prog1: Student Information Display

name = input("Enter student's name: ")
age = input("Enter student's age: ")
course = input("Enter student's course: ")

print("\n--- Student Information ---")
print("Name:", name)
print("Age:", age)
print("Course:", course)
```

```python
# fa3prog2: Pass or Fail Checker

score = float(input("Enter exam score: "))

if score >= 75:
    print("Passed")
else:
    print("Failed")
```

```python
# fa3prog3: Grading System

name = input("Enter student's name: ")
score = float(input("Enter student's score: "))

if 90 <= score <= 100:
    grade = "A"
elif 80 <= score <= 89:
    grade = "B"
elif 70 <= score <= 79:
    grade = "C"
else:
    grade = "F"

print("\n--- Result ---")
print("Name:", name)
print("Grade:", grade)
```
